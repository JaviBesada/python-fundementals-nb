# Diccionarios

Los items se almacenan y recuperar por una clave, en vez de por su posición en la estructura.

Es una tipo de agregación en el cual el nombre de los items tiene más significado que su posición.

Los diccionarios pueden reemplazar muchos de los algoritmos de búsqueda y estructuras de datos que deberías implementar manualmente en lenguajes de bajo nivel.

***Características:***
- Se accede a los elementos por su clave o key, no por la posicion que ocupan.
- Son colecciones no ordenadas de objetos de cualquier tipo, y el tipo de los elementos que lo forman puede ser distinto.
- Poseen longitud variable.
- Son mapas mutables.
- Son tablas de referencias a objetos o hash tables.

### Metodos de un objeto diccionario

#### values

El método **values** devuelve todos los valores del diccionario.

```python
D = {"spam": 2, "ham": 1, "eggs": 3}
list(D.values())
```

#### items

**items** devuelve todas las tuplas **(key, value)**.

```python
list(D.items())
D
```

#### get

Buscando una clave que no existe es normalmente un error,, pero el metodo **get** devuelve **None** como valor por defecto o un valor pasado como por defecto si la clave no existe.

```python
D["toast"]
```

```python
D.get("spam")
```

```python
print(D.get("toast"))   # Clave que no existe.
```

```python
D.get("toast", 88)      # Añadiendole un valor por defecto.
```

#### update

El metodo **upadate** une las claves y valores de un diccionario en otro, sobreescribiendo valores de la misma clave si hay conflicto.

```python
# D = {"eggs": 3, "spam": 2, "ham": 1}
D2 = {"toast": 4, "muffin": 5, "ham": 10}
D.update(D2)
D
```

#### pop

**pop** elimina una clave de un diccionario y devuelve el valor asociado a la clave.

```python
D.pop("muffin")
# 5
D.pop("toast")
# 4
D
# {"eggs": 3, "spam": 2, "ham": 10}
```

### Ejemplos

Movie Database

```python
table = {'1975': 'Holy Grail', '1979': 'Life of Brian', '1983': 'The Meaning of Life'}
year = "1983"
movie = table[year]
movie
```

```python
for year in table:                      # Lo mismo que: for year in table.keys()
    print(year + "\t" + table[year])
```

Para cualquier diccionario D, **for key in D** funciona de la misma manera que **for key in D.keys()**

```python
for key in table:
    print(key + "\t" + table[key])
```

```python
for key in table.keys():
    print(key + "\t" + table[key])
```

Mapeando valores a claves:

```python
table = {'Holy Grail': '1975', 'Life of Brian': '1979', 'The Meaning of Life': '1983'}
table["Holy Grail"]
```

```python
list(table.items())     # Lista de tuplas(clave, valor)
```

#### Comprension de sintaxis

```python
[title for (title,year) in table.items() if year == 1975]     #Titulo por titulo,año en los items de la table si el año es 1975
```

```python
V = "1975"
[key for (key, value) in table.items() if value == V]    
```

```python
[key for key in table.keys() if table[key] == V]    # Coge cada clave en la tabla clave, y luego si tabla[key] devuelve el valor de esa key, y comprueba si el valor es igual a V y lo devuelve
```

### Como crear diccionarios

```python
{"name": "Bob", "age": 40}      # Expresion literal
```

```python
D = {}          # Asignacion dinamica de claves
D["name"] =  "Bob"
D["age"] = 40
```

```python
dict(name = "Bob", age = 40)        # Forma dict keyword argument
```

En el **zip** cogemos una lista o tupla (keylist) metemos las claves ("name", "age") y luego en otra lista o tupla metemos los valores ("Bob, 40), la funcion **zip** lo que hace es cogerlo por orden y combinar la clave (por orden) por ejemplo coge "name" como clave y le mete bob como valor.

```python
keylist = ["name", "age"]
valuelist = ["Bob", 40]
dict(zip(keylist, valuelist))       #Forma Zipped key/value tuples.
```

```python
dict.fromkeys(["a", "b"], 0)
```

### Vistas diccionario y sets

La vista obeto devuelta por el metodo **keys** es como un **set** y soporta las operaciones con conjuntos, como la interseccion y la union

```python
K, V
```

```python
K | {"x": 4}        # La vista `keys` es como un `set`
```

La lista **values** no es como un **set**.

```python
V & {"x": 4}     # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict'
```

La vista **items** genera pares (key, value) que son unicos e inmutables.

```python
V & {'x': 4}.values()   # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict_values'
```

### Ordenando clave

```python
Ks = D.keys()       # Obtenemos las claves en una variable.
Ks = list(Ks)       # Las transformamos en una lista.
Ks.sort()           # Las ordenamos
for k in Ks:        # Las vamos buscando
    print(k, D[k])  # Las printeamos
```

```python
D = {"b": 2, "c": 3, "a": 1}        # Creamos el diccionario.
Ks = D.keys()                       # Añadimos a la variable Ks las claves
for k in sorted(Ks):                # Por clave en calves ordenadas
    print(k, D[k])                  # Printeamos
```
