## Ambitos globales

- Los nombres globales son variables a las que se asigna un valor en el nivel superior del archivo del módulo que las contiene.
- Los nombres globales solo deben declararse si se les asigna un valor dentro de una función.
- Se puede hacer referencia a los nombres globales dentro de una función sin necesidad de declararlos.



```python
x = 88      # Global X

def func():
    x = 99      # Local X: oculta la global X pero queremos que esta este aqui.

func()
print(x)  # Imprime 88, la variable global X no se ve afectada por la local X dentro de func(), porque la variable local x solo existe dentro del alcance de la función func().
```

```python
x = 88      # Global X

def func():
    global x  # Declaramos que queremos usar la variable global X dentro de la función.
    x = 99      # Modifica la variable global X. Que esta fuera de la funcion.

func()
print("x =", x)  # Imprime 99, la variable global X ha sido modificada por la función func().
```

## LEGB rule

```python
# Se puede hacer referencia a nombres globales dentro de una función sin necesidad de declararlos.

y, z = 1, 2         # Variables globales en el modulo

def all_global():
    global x        # Declaramos que queremos usar la variable global X dentro de la función.
    x = y + z       # No es necesario declarar y y z como globales, ya que solo estamos haciendo referencia a ellas, no modificándolas.
                    # X existe ahora en el ambito global con valor 3.

all_global()
print ("x = %d, y = %d, z = %d" % (x, y, z))  # Imprime x = 3, y = 1, z = 2. %d hace referencia a un entero, y %s hace referencia a una cadena de texto.
```

## Nested scopes

```python
x = 99          # Global X no se ve afectada por la local X dentro de f1() y f2(), porque la variable local x solo existe dentro del alcance de la función f1().

def f1():
    x = 88      # Local X: oculta la global X pero queremos que esta este aqui.
    def f2():   # Incluyendo def local f2() anidada dentro de f1().   
        print("x local=", x)    # Referencia de una def anidada.
    f2()        # f2 es una función temporal que solo existe durante la ejecución de f1 (y solo es visible para el código de esta última)

f1()            # Prints 88, la variable local x dentro de f1() es visible para la función anidada f2().    
print("x global=", x)  # Imprime x global= 99, la variable global X no se ve afectada por la local X dentro de f1() y f2(), porque la variable local x solo existe dentro del alcance de la función f1().

# Si pusiesemos f2()  NameError: el nombre «f2» no está definido. No se puede llamar a f2() desde fuera de f1(), ya que f2() es una función local definida dentro de f1() y no es accesible desde el ámbito global.
```

## Fuenciones de fábrica: Cierres

Una *closure* es una función que recuerda las variables de la función donde se creó, incluso cuando esa función ya ha terminado. Una *factory function* es una función que crea y devuelve otra función u objeto para poder utilizarlo después.

```python
def f1():
    x = 88      
    def f2():   # Recuerda la variable x en el ámbito de la definición que la contiene
        print("x =", x)   # Devuelve f2, pero no la llama => f1() devuelve el objeto función con el nombre (referencia) f2
    return f2    # Devuelve f2, pero no la llama => f1() devuelve el objeto función con el nombre (referencia) f2

action = f1()  # action es ahora una referencia a la función f2() que fue devuelta por f1()
action()       # LLama a la función f2() a través de la referencia action, que recuerda el valor de x en el ámbito de f1().

# En Python, las funciones son objetos como cualquier otra cosa y pueden devolverse como valores de retorno desde otras funciones.
# Y lo más importante: f2 recuerda la variable x del ámbito circundante en f1, aunque f1 ya no esté activa.
```

```python
def maker(n):
    x = 88
    def action(x):  # Accion de creación y devolucion.
        return x ** n  # La acción conserva la variable «n» del ámbito circundante
    return action  

f = maker(2) # Pasa 2 al argumento n -> devuelve x ** 2
f(3) # Se pasa 3 a x, n recuerda «2» -> devuelve 3 ** 2

f(4) # Devuelve 4 ** 2
```
