## Objetos mutables

### Listas

```python
L = [2, 3, 4]   # Un objeto mutable
M = L           # Crear una referancia al objeto.
L[0] = 24       # Un cambio in situ. Aqui si cambia porque L y M apuntan al la misma lista en memoria.

print("L =", L)
print("M =", M)
```

Copia de objetos en vez de referencia:

```python
L = [2, 3, 4]
M = L[:]        # Crear una copia de la lista. Tambien se puede hacer con list(L) o L.copy().
L[0] = 24       # Un cambio in situ. Aqui no cambia porque L y M apuntan a diferentes listas en memoria.

print("L =", L)

print("M =", M) # M sigue siendo la lista original.
```

### Referencias compartidas e igualdad
    
    - Operador *==* igualdad de valores.
    - Operador *is* identidad de objetos (es). 

```python
L = [1, 2, 3]
M = L           # Crear una referancia al objeto.

print("L == M es", L == M) # True, porque tienen el mismo contenido.

print("L is M es", L is M) # True, porque son el mismo objeto.
```

```python
L = [1, 2, 3]
M = [1, 2, 3]       # M y L sos dos listas diferenetes.

print("L == M es", L == M) # True, porque tienen el mismo contenido.

print("L is M es", L is M) # False, porque son dos objetos diferentes
```

### Cache

Sin embargo, dado que los números enteros pequeños y las cadenas se almacenan en caché y se reutilizan, esto nos indica que hacen referencia al mismo objeto.

```python
X = 42
Y = 42      #Deberian ser dos objetos diferentes, pero Python optimiza el uso de memoria y reutiliza el mismo objeto para valores pequeños de enteros.

X == Y # True, porque tienen el mismo contenido.
X is Y # True, porque son el mismo objeto.
```

Averiguar el numero de referencias a un objeto:

```python
import sys
sys.getrefcount(X) # Devuelve el numero de referencias al objeto X.
```
