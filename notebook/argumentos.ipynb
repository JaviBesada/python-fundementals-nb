## 1.Argumentos y referencias compartidas

![Imagen Pulpo](../img/Captura%20de%20pantalla%20de%202026-10-21%2012-50-48.png)

### Paso de objetos inmutables:

```python
def f(a):     # Se asigna a (que hace referencia a) al objeto pasado
    a = 99    # Solo se modifica la variable local a

b = 88 
f(b)          # Inicialmente, tanto a como b hacen referencia al mismo 88
print("b= ", b) # Inicialmente, tanto a como b hacen referencia al mismo 88
```

### Paso de objetos mutables:

```python
def changer(a, b):  # Los argumentos tienen asignadas referencias a objetos
    a = 2           # Solo cambia el valor de la variable local
    b[0] = "spam"     # Modifica el objeto compartido in situ

x = 1
l = [1,2]
changer(x, l)       # Llamada: se pasan objetos inmutables y mutables
print("l = ", l)    # ['spam', 2]  # ¡x no ha cambiado, pero l sí!
assert x == 1, "x es 1" #Se asegura que x es igual a 1, si es cierto sigue, si no da error.
```

### Cómo evitar que una rutina modifique los elementos mutables:

-Pasar una copia

```python
l = [1, 2]
changer(x, l[:]) 	# Pasa una copia y l no cambia.

print("l = ", l) 
```

-Utilitar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]        # Copiamos la lista de entrada para no afectar a la función que la llama
    a = 2
    b[0] = "spam"   # Solo modificamos nuestra copia de la lista

changer(x, l)
assert l == [1, 2], "l es una invariante" #Es una invariante, debe ser asi, por ejemplo el sueldo nunca debe ser 0. Quien lo quiera hacer sabe que esto no puede cambiar ni variar y esto se asegura de ello.
print("l = ", l)
```

-Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]
def changer(a, b):
    a = 2
    b[0] = "spam"       # TypeError: el objeto 'tuple' no admite la asignación de elementos

changer(x, tuple(l))    # Al pasar una tupla, los cambios se consideran errores, porque las tuplas son inmutables.

print("l = ", l)
assert l ==  [1, 2], "l es una invariante"
```

## 2. Modos especiales de coincidencia de argumentos

Ver figura 06 en drive

Hay que tener cuidado de no confundir nombre=valor en el encabezado de una función y en una llamada, porque se escribe igual pero significa cosas distintas. En el encabezado ***(def saludar(nombre, saludo="Hola")), saludo="Hola"*** define un valor por defecto: si quien llama no da ese argumento, la función usa "Hola", así que el argumento es opcional. En la llamada ***(saludar(nombre="Ana", saludo="Buenas"))***, en cambio, nombre=valor es un argumento de palabra clave: indica a qué parámetro va cada valor usando su nombre en lugar de su posición, por lo que el orden ya no importa.

### 1.Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)          # 1 2 3
```

### 2.Keywords

El orden de los argumentos de izquierda a derecha ya no importa cuando se utilizan palabras clave, ya que los argumentos se emparejan por su nombre, no por su posición.

```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

### 3.Defaults

Los valores por defecto nos permiten hacer que determinados argumentos de una función sean opcionales; si no se les pasa ningún valor, a la función se le asigna su valor por defecto antes de que se ejecute.

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```

Si pasamos dos valores, solo c toma su valor por defecto, y si pasamos tres valores, no se utiliza ningún valor por defecto:

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)
    
f(1, 4)     # 1 4 3
f(1, 4, 5)	# Anular los valores por defecto: 1 4 5

f(1, c=6)	# Elegir los valores por defecto: 1 2 6
```

### 4. Combining keywords and defaults

Fíjate de nuevo en que, cuando se utilizan argumentos por palabra clave en la llamada, el orden en que aparecen los argumentos no importa; Python los identifica por su nombre, no por su posición.

El llamante debe proporcionar valores para «spam» y «eggs», pero estos pueden asignarse tanto por posición como por nombre.

```python
def func(spam, eggs, toast=0, ham=0):
	print((spam, eggs, toast, ham)) 	# First 2 required

func(1, 2)							# Output: (1, 2, 0, 0)
func(1, ham=1, eggs=0)				# Output: (1, 0, 0, 1)
func(spam=1, eggs=0)				# Output: (1, 0, 0, 0)
func(toast=1, eggs=2, spam=3)		# Output: (3, 2, 1, 0)
func(1, 2, 3, 4) 					# Output: (1, 2, 3, 4)
```

### 5. Python 3.X Keyword-Only Arguments

Un * suelto en la definición de una función funciona como una barrera: todo lo que va después de él solo se puede pasar por palabra clave (con nombre=valor), nunca por posición.

```python
def kwonly(a, *b, c):
    print(a, b, c)

kwonly(1, 2, c=3)       # 1 (2,) 3

kwonly(a=1, c=3)        # 1 () 3

kwonly(1, 2, 3)         kwonly(1, 2, 3)         # TypeError: a kwonly() le falta un argumento obligatorio que solo puede ser una palabra clave: «c»
```

Aún así, puedes utilizar valores por defecto para los argumentos que solo admiten palabras clave, aunque aparezcan después del * en la declaración de la función:

```python
def kwonly(a, *, b='spam', c='ham'):
	print(a, b, c)

kwonly(1)               # 1 spam ham
kwonly(1, c=3)          # 1 spam 3
kwonly(a=1)             # 1 spam ham
kwonly(c=3, b=2, a=1)   # 1 2 3
kwonly(1, 2)            # TypeError: kwonly() admite 1 argumento posicional, pero se han proporcionado 2
```

```python
def kwonly(a, *, b, c='spam'):
    print(a, b, c)

kwonly(1, c='eggs')     # TypeError: a kwonly() le falta un argumento obligatorio de tipo «keyword-only»: «b»
```

## 3.Arbitrary Arguments Examples

Las dos últimas extensiones de coincidencia, * y **, están diseñadas para admitir funciones que aceptan cualquier número de argumentos.

Ambas pueden aparecer tanto en la definición de la función como en una llamada a la función.

### Headers: Collecting arguments

En la definición de la función, recopila los argumentos posicionales sin correspondencia en una tupla y asigna la variable `args` a dicha tupla.

```python
def f(*args):  # args es para poder meter muchos argumentos. Cuando ponemos * los convierte en una tupla por tanto son inmutables.
    print(args)

f()                 # ()
f(1,)               # (1,)
f(1, 2, 3, 4)       # (1, 2, 3, 4)  
```

```python
La función ** solo funciona con argumentos de palabra clave. Los agrupa en un nuevo diccionario.
```

```python
def f(**args):
    print(args)
f()                 # {}
f(a=1, b=2)         # {'a': 1, 'b': 2} # Necesitan ser palabras clave a = 1 
```

```python
def f(a, *pargs, **kargs):
    print(a, pargs, kargs)
f(1, 2, 3, x=1, y=2)        # 1 (2, 3) {'y': 2, 'x': 1}
```

### Calls: Unpacking arguments

  "*" La sintaxis en una función de llamada desempaqueta un conjunto de argumentos.

```python
def func(a, b, c, d): 
    print(a, b, c, d)

args = (1, 2)
args += (3, 4)
func(*args)         # Same as func(1, 2, 3, 4)
```

```python
def func(a, b, c, d): 
    print(a, b, c, d)
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # Same as func(a=1, b=2, c=3, d=4)
func(*args)
```

```python
func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```
