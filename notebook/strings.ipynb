## Polimorfismo

El signo más (+) tiene diferentes significados según el objeto: suma en el caso de los números y concatenación en el de las cadenas.

Esta es una propiedad general de Python denominada polimorfismo.

El significado de una operación depende de los objetos sobre los que se realiza.

Como verás cuando estudiemos el tipado dinámico, esta propiedad del polimorfismo explica en gran medida la concisión y la flexibilidad del código de Python.

Dado que los tipos no están restringidos, una operación escrita en Python normalmente puede funcionar automáticamente con muchos tipos diferentes de objetos, siempre que estos admitan una interfaz compatible (como la operación + en este caso).

## Inmutabilidad

Los objetos inmutables no se pueden modificar.

Todos los objetos en Python se clasifican como inmutables (que no se pueden modificar) o mutables.

En cuanto a los tipos básicos:

los números, las cadenas y las tuplas son inmutables;
las listas, los diccionarios y los conjuntos son mutables.

```python
S[0] = "z"  # Esto no funciona, las cadenas son inmutables
```

## Métodos específicos de tipo

### Find(encontrar)

```python
# Hallar la posición de una subcadena en S. Encuentra la primera aparición de la subcadena y devuelve el índice de su primer carácter. Si no se encuentra, devuelve -1.
S = "SpamEggsSpam"
S.find("pa")
```

```python
S.find("pa", 3)  # Busca "pa" a partir del desplazamiento 3
```

```python
S.find(S) # Busca la cadena S dentro de S, lo cual devuelve 0, ya que la cadena se encuentra a partir del índice 0.
```

```python
# Si no existe el caracter => -1
S.find('z')
```

### Replace(sustituir)

```python
S.replace("Eggs", "Bacon")  # Reemplaza todas las apariciones de "Eggs" con "Bacon"
# S es inmutable, por lo que no cambia. Si quieres guardar el resultado, debes asignarlo a una variable.
```

```python
S #Sigue siendo igual aunque hayamos hecho un replace, ya que las cadenas son inmutables. Si quieres guardar el resultado, debes asignarlo a una variable.
```

```python
S.upper()  # Devuelve una nueva cadena con todos los caracteres en mayúsculas
# S sigue siendo igual, ya que las cadenas son inmutables. Si quieres guardar el resultado, debes asignarlo a una variable.
```

```python
print("S is alpha", S.isalpha())  # Comprueba si todos los caracteres de S son alfabéticos
print("S is digit", S.isdigit())  # Comprueba si todos los caracteres de S son dígitos
```

### Split(dividir)

```python
S = 'spams-spam-eggs-spam-bacon'
S.split('-') #Separa la cadena en una lista de subcadenas.
```

## Rstrip

Elimina los caracteres de espacio en blanco a la derecha ("/n" -> salto de linea, "/t" -> tabulación)

```python
line = 'aaa\t,bbb\t,\nccccc,\tdd\n' #Esto es un ejemplo de una cadena que contiene tabulaciones y saltos de línea.
print(line)
```

```python
line.rstrip()  # Elimina los espacios en blanco a la derecha de la cadena
```

```python
# Combinación de dos oporedores
line.rstrip().split(',')  # Primero elimina los espacios en blanco a la derecha y luego separa la cadena en una lista de subcadenas usando la coma como delimitador.
```

## Formating(formateo)

Las cadenas también admiten una operación de sustitución avanzada conocida como ***formateo***

```python
# Expresion de formateo(formating expression)
"%s, eggs, and %s" % ("spam", "SPAM!")  # Esto es un ejemplo de formateo de cadenas usando el operador %.
```

```python
# Metodo de formateo(formating method)
"{}, eggs, and {}".format("spam", "SPAM!")  
```

```python
S = "SpamEggsSpam"
f"{S}" #Esto es un ejemplo de formateo de cadenas usando f-strings, que es una forma más moderna y legible de formatear cadenas en Python.
```

```python
'{0}'.format(S) #Esto es un ejemplo de formateo de cadenas usando el método format() con índices.
```

## Herramientas de conversión de cadenas

```python
"42" + 1 #Esto no funciona, ya que estamos intentando concatenar una cadena con un entero.
```

```python
# SUMA
int("42") + 1 #Esto funciona, ya que estamos convirtiendo la cadena "42" a un entero antes de sumarle 1.
```

```python
# CONCATENACION
"42" + str(1) #Esto funciona, ya que estamos convirtiendo el entero 1 a una cadena antes de concatenarlo con la cadena "42".
```

```python
# La función `repr` convierte un objeto en su representación como cadena
repr(42) # Esto devuelve la representación de la cadena del número entero 42, que es '42'.
```

```python
repr("42") #Esto devuelve la representación de la cadena "42", que es "'42'".
```

## Conversiones de códigos de caracteres

Para convertir un carácter concreto en su código entero subyacente (por ejemplo, ***su valor en bytes ASCII***), hay que pasarlo a la función integrada `ord`, que devuelve el valor binario real utilizado para representar el carácter correspondiente en memoria.

En el caso del texto ASCII, se trata del conocido entero de 7 bits que cabe en un solo byte en memoria, pero el rango de puntos de código para otros tipos de texto ***Unicode*** puede ser más amplio.

```python
ord('s') #Este es un ejemplo de la función `ord`, que devuelve el valor Unicode del carácter 's', que es 115.
```

La función `chr` realiza la operación inversa: toma un código entero y lo convierte en el carácter correspondiente:

```python
chr(115) #Este es un ejemplo de la función `chr`, que devuelve el carácter correspondiente al valor Unicode 115, que es 's'.
```
